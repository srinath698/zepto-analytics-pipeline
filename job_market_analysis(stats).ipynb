```python
#application of inferenctial stats
import pandas as pd
import numpy as np

df = pd.read_csv("ai_jobs.csv")
sample = df.sample(n=500, random_state=42)

average_salary_max = sample["salary_max_usd"].mean()
average_salary_min = sample["salary_min_usd"].mean()
average_salary = (average_salary_max + average_salary_min) / 2

print("Estimated Average Salary:", average_salary)
```

```python
import numpy as np

population = np.array([12, 15, 18, 20, 22, 24, 25, 28, 30, 35])

print(population)
```

```python
pop_mean=np.mean(population)
print('pop_mean',pop_mean)
```

```python
pop_var=np.var(population)
print('pop_var',pop_var)
```

```python
pop_std=np.std(population)
print('pop_std',pop_std)
```

```python
high_salary=population>=25
pop_prop=np.mean(high_salary)
print('pop_prop',pop_prop)
```

```python
sample = np.array([15, 20, 24, 28, 35,44,55,67])

print(sample)
```

```python
sample_mean=np.mean(sample)
print('sample_mean',sample_mean)
```

```python
sample_var=np.var(sample)
print('sample_var',sample_var)
```

```python
sample_std=np.std(sample)
print('sample_std',sample_std)
```

```python
sample_prop=np.mean(sample>=25)
print('sample_prop',sample_prop)
```

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("ai_jobs.csv")
salary = df["salary_min_usd"]

ci = stats.t.interval(
    confidence=0.95,
    df=len(salary)-1,
    loc=salary.mean(),
    scale=stats.sem(salary)
)

print(ci)
```
